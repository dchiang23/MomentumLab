```python
import pandas as pd #tables / data
import yfinance as yf #historical data 
import matplotlib.pyplot as plt #plotting data
import numpy as np #numerical operations

SPY_data = yf.download("SPY", "2010-01-01", "2026-09-01", interval = "1d", auto_adjust=False)
QQQ_data = yf.download("QQQ", "2010-01-01", "2026-09-01", interval = "1d", auto_adjust=False)
IWM_data = yf.download("IWM", "2010-01-01", "2026-09-01", interval = "1d", auto_adjust=False)

SPY_data.head(5) #shows the first 5 rows of the dataframe
SPY_data.tail(5) #shows the last 5 rows of the dataframe -> overwrites the previous call in terms of showing output
SPY_data.shape #tells me the number of rows, columns
SPY_data.columns #shows the names of the variables/columns
SPY_data.index ## shows the row labels, usually trading dates
```

```python
print(SPY_data.index.is_monotonic_increasing) #is it sorted chronologically
print(SPY_data.index.has_duplicates) #checks for duplicate dates
#check if contains any non trading days
(SPY_data.index.dayofweek >= 5).any()
```

```python
#check for rmissing values
SPY_data.isna().sum()
QQQ_data.isna().sum()
```

```python
SPY_data.head(5)
#adjusted close is closing price adjusted for events that change the apparent price without representing an 
#actual investment loss / gain
#so like stock splits, dividends
```

```python
##Plotting Practice:
plt.plot(SPY_data.index, SPY_data['Adj Close'], label = "SPY")
plt.plot(QQQ_data.index, QQQ_data['Adj Close'], label = "QQQ")
plt.plot(IWM_data.index, IWM_data['Adj Close'], label = "IWM")
plt.legend()
plt.xlabel("Time")
plt.ylabel("Adj Close")
plt.title("Adjusted Close Prices: SPY, QQQ, and IWM")
plt.grid()
plt.show()
```
